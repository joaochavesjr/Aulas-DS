```python
import pandas as pd
```

```python
df = pd.read_csv('quadro_medalhas.csv')
```

```python
ouro = df[df['Medalha']=='Ouro']
```

```python
ouro_por_ano = ouro.groupby('Ano')['Quantidade'].sum()
```

**1. Gráfico de linhas (line): quantidade de medalhas de ouro ao longo dos anos.**

```python
ouro_por_ano.plot(kind='line', title='Quadro de Medalhas Total por Ano', ylabel='Quantidade', xlabel='Ano')
```

```python
ouro_por_pais = ouro.groupby('Pais')['Quantidade'].sum()
```

**2. Gráfico de barras verticais (bar): quantidade de medalhas de ouro por país**

```python
ouro_por_pais.plot(kind="bar", title='Total Ouro por País', ylabel='Qtde', x='País')
```

**3. Gráfico de barras horizontais (barh): quantidade de medalhas de ouro por país**

```python
ouro_por_pais.plot(kind="barh", title='Total Ouro por País', ylabel='Qtde', x='País')
```

**4. Histograma (hist): distribuição da quantidade de medalhas de ouro**

```python
ouro_por_pais.plot(kind="hist", bins=3, title="Distribuição das Medalhas de Ouro por País", xlabel="Quantidade de Ouro", ylabel="Número de Países")
```

**5. Boxplot (box): distribuição da quantidade de medalhas de ouro.**

```python
ouro_por_pais.plot(kind='box', title='Medalhas de Ouro por País')
print(ouro_por_pais.sort_values(ascending=False))
```

**6. Gráfico de densidade de kernel (kde): distribuição da quantidade de medalhas
de ouro.**

```python
ouro_por_pais.plot(kind='kde', title='Quantidade Medalhas Ouro', xlabel='Qtde', ylabel='Densidade')
```

**7. Gráfico de densidade (density): distribuição da quantidade de medalhas de ouro.**

```python
ouro_por_pais.plot(kind='density', title='Quantidade de Medalhas', xlabel='Qtde', ylabel='Densidade')
```

**8. Gráfico de área (area): quantidade de medalhas de ouro ao longo dos anos.**

```python
ouro_ano_pais = ouro.groupby(['Ano', 'Pais'])['Quantidade'].sum().unstack()
ouro_ano_pais.plot(kind='area', title='Quadro Medalhas por Ano', xlabel='Ano', ylabel='Qtde')
```

**9. Gráfico de pizza (pie): proporção de medalhas de ouro por país.**

```python
ouro_por_pais.plot(kind='pie', title='Medalhas de Ouro por País',
                   autopct='%1.1f%%',
                   ylabel="")
```

**10. Gráfico de dispersão (scatter): quantidade de medalhas de ouro por ano.**

```python
ouro_por_ano = ouro.groupby('Ano')['Quantidade'].sum().reset_index()
ouro_por_ano
```

```python
ouro_por_ano.plot(kind='scatter', x='Ano', y='Quantidade', title='Quantidade de Medalhas por Ano', xlabel='Ano')
```
